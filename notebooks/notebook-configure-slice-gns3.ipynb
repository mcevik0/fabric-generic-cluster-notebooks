## Step 4: Install GNS3


### Select the node to install GNS3

```python
node_name="lc-1"
```

### Upload scripts to the node

```python
node = slice.get_node(name=node_name)

try:
    node.upload_file('../tools/node_tools/install_wireshark.sh', 'install_wireshark.sh')
    node.upload_file('../tools/node_tools/install_docker.sh', 'install_docker.sh')
    node.upload_file('../tools/node_tools/install_gns3.sh', 'install_gns3.sh')
except Exception as e:
    print(f"Exception: {e}")
```

### Install

```python
try:
    stdout, stderr = node.execute(f'chmod +x install_wireshark.sh && sudo ./install_wireshark.sh > /tmp/install_wireshark.log 2>&1')
except Exception as e:
    print(f"Exception: {e}")
```

```python
try:
    stdout, stderr = node.execute(f'chmod +x install_docker.sh && sudo ./install_docker.sh > /tmp/install_docker.log 2>&1')
except Exception as e:
    print(f"Exception: {e}")
```

```python
try:
    stdout, stderr = node.execute(f'chmod +x install_gns3.sh && sudo ./install_gns3.sh > /tmp/install_gns3.log 2>&1')
except Exception as e:
    print(f"Exception: {e}")
```

```python
reboot = 'sudo reboot'
try:
    print(reboot)
    node.execute(reboot)
    
    slice.wait_ssh(timeout=360,interval=10,progress=True)

    print("Now testing SSH abilites to reconnect...",end="")
    slice.update()
    slice.test_ssh()
    print("Reconnected!")

except Exception as e:
    print(f"Fail: {e}")  

node.config()
print("Done")    
```
