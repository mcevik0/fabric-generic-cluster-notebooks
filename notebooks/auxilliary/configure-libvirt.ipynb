## Step 4: Install Libvirt


### Select the node to install Libvirt

```python
node_name="Node1"
```

### Upload scripts to the node

```python
node = slice.get_node(name=node_name)

try:
    node.upload_file('../../tools/node_tools/install_libvirt.sh', 'install_libvirt.sh')
    node.upload_file('../../tools/node_tools/create_vm.sh', 'create_vm.sh')
except Exception as e:
    print(f"Exception: {e}")
```

### Install

```python
try:
    stdout, stderr = node.execute(f'chmod +x install_libvirt.sh && sudo ./install_libvirt.sh > /tmp/install_libvirt.log 2>&1')
except Exception as e:
    print(f"Exception: {e}")
```

```python
reboot = 'sudo reboot'
try:
    print(reboot)
    node.execute(reboot)
    
    slice.wait_ssh(timeout=360,interval=10,progress=True)

    print("Now testing SSH abilites to reconnect...",end="")
    slice.update()
    slice.test_ssh()
    print("Reconnected!")

except Exception as e:
    print(f"Fail: {e}")  

node.config()
print("Done")    
```

```python
try:
    stdout, stderr = node.execute(f'chmod +x create_vm.sh && sudo ./create_vm.sh > /tmp/create_vm.log 2>&1')
except Exception as e:
    print(f"Exception: {e}")
```
