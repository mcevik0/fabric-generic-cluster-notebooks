## Step 3: Install Required Packages

### Option A: Install

````python
# Install 
import os
import shutil
from pathlib import Path

def install_fabric_cluster(method="pypi"):
    """
    Install fabric-generic-cluster using either:
      - method="pypi"
      - method="local"
    """

    method = method.lower()
    repo_dir = Path("fabric-generic-cluster")

    if method not in ("pypi", "local"):
        print(f"[ERROR] Unknown method: {method}")
        return

    if method == "pypi":
        print("[INFO] Installing from PyPI...")
        !pip install --upgrade fabric-generic-cluster --quiet
        print("[DONE] Installed from PyPI.")
        return

    if method == "local":
        print("[INFO] Installing from local git repository...")

        # Remove any previous clone
        if repo_dir.exists():
            shutil.rmtree(repo_dir)

        # Clone fresh copy
        !git clone https://github.com/mcevik0/fabric-generic-cluster.git

        # Install editable version
        current = os.getcwd()
        os.chdir(repo_dir)

        try:
            !pip install -e .
        finally:
            os.chdir(current)

        print("[DONE] Installed from local git clone.")
        return
````

## Step 4: Verify Installation

Check if all required packages are installed correctly:

````python
import sys

def check_package(package_name, import_name=None):
    """Check if a package is installed and importable."""
    if import_name is None:
        import_name = package_name
    
    try:
        module = __import__(import_name)
        version = getattr(module, '__version__', 'unknown')
        print(f"✅ {package_name:20s} {version}")
        return True
    except ImportError:
        print(f"❌ {package_name:20s} NOT INSTALLED")
        return False

print("\n" + "="*60)
print("Package Installation Status")
print("="*60 + "\n")

# Core packages
print("Core Packages:")
check_package('pydantic')
check_package('PyYAML', 'yaml')

# Visualization
print("\nVisualization Packages:")
check_package('matplotlib')
check_package('networkx')

# Utilities
print("\nUtility Packages:")
check_package('tabulate')

# Optional packages
print("\nOptional Packages:")
check_package('pydot')
check_package('pydotplus')

# FABRIC (if needed)
print("\nFABRIC Testbed:")
check_package('fabrictestbed_extensions', 'fabrictestbed_extensions')

print("\n" + "="*60)
````

## Troubleshooting

### Issue: Package not found after installation

Try restarting the kernel:
- Kernel → Restart Kernel

### Issue: Permission denied

Try installing with user flag:
```bash
pip install --user -r requirements.txt
```

### Issue: Conflicting package versions

Create a virtual environment:
```bash
python -m venv venv
source venv/bin/activate  # On Windows: venv\Scripts\activate
pip install -r requirements.txt
```

### Issue: fabrictestbed-extensions conflicts

If `fabrictestbed-extensions` is already installed in your JupyterHub environment, you may need to comment it out in `requirements.txt`.
